```python
import seaborn as sns
```

```python
#load_dataset():loads a bult-in dataset
tips = sns.load_dataset("tips")
iris = sns.load_dataset("iris")
flights = sns.load_dataset("flights")
```

```python
import matplotlib.pyplot as plt

#sns.scatterplot():draws a scatter plot showing the relationship between two numerical variables
sns.scatterplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
#lineplot():it draws a line chart 
sns.lineplot(data=flights, x="year", y="passengers")
plt.show()
```

```python
#relplot():figure level function for scatter and line plots
sns.relplot(data=tips, x="total_bill", y="tip", kind="scatter")
```

```python
#barplot():displays averages using bars
sns.barplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
#countplot():counts the occurrences of categories 
sns.countplot(data=tips, x="day")
plt.show()
```

```python
#pointplot():Shows point estimates and confidence intervals.
sns.pointplot(data=tips, x="day", y="tip")
plt.show()
```

```python
#catplot():Figure-level interface for categorical plots
sns.catplot(data=tips, x="day", y="tip", kind="box")
```

```python
#boxplot():Displays quartiles and detects outliers
sns.boxplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
#violinplot():shows data distribution and density
sns.violinplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
#boxenplot():Enhanced boxplot for large datasets.
sns.boxenplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
#stripplot():jittered scatter across categories.
sns.stripplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
#swarmplot():Displays individual observations without overlap.
sns.swarmplot(data=tips, x="day", y="total_bill")
plt.show()
```

```python
#histplot():draws the histogram
sns.histplot(data=tips, x="total_bill")
plt.show()
```

```python
#kdeplot():Kernel Density Estimation plot.
sns.kdeplot(data=tips, x="total_bill")
plt.show()
```

```python
#ecdfplot():Empirical cumulative distribution.
sns.ecdfplot(data=tips, x="total_bill")
plt.show()
```

```python
#rugplot():Displays individual observations as small ticks.
sns.rugplot(data=tips, x="total_bill")
plt.show()
```

```python
#displot():Figure-level distribution plot.
sns.displot(data=tips, x="total_bill")
```

```python
#jointplot():Shows relationship plus distributions.
sns.jointplot(data=tips, x="total_bill", y="tip")
```

```python
#pairplot():Plots pairwise relationships.
sns.pairplot(iris)
```

```python
#PairGrid():custom pairwise plots
g = sns.PairGrid(iris)
g.map(sns.scatterplot)
```

```python
#lmplot():Regression line with scatter.
sns.lmplot(data=tips, x="total_bill", y="tip")
```

```python
#regplot():regressing plot
sns.regplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
#residplot():residual plot
sns.residplot(data=tips, x="total_bill", y="tip")
plt.show()
```

```python
#heatmap():Displays matrix values using colors.
corr = iris.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.show()
```

```python
%pip install scipy

import scipy

#clustermap():heatmap with hierarchical clustering.
sns.clustermap(corr)
plt.show()
```

```python
#FacetGrid():Creates multiple plots by category.
g = sns.FacetGrid(tips, col="time")
g.map(sns.histplot, "total_bill")
```

```python
#JointGrid():Custom joint plots
g = sns.JointGrid(data=tips, x="total_bill", y="tip")
g.plot(sns.scatterplot, sns.histplot)
```

```python
#objects.Plot():Modern object-oriented plotting interface
from seaborn.objects import Plot

(
    Plot(tips, x="total_bill", y="tip")
    .add(sns.objects.Dots())
)
```

```python
#move_legend():Moves the legend.
ax = sns.scatterplot(data=tips, x="total_bill", y="tip", hue="day")
sns.move_legend(ax, "upper left")
plt.show()
```

```python
#despine():Removes chart borders.
sns.scatterplot(data=tips, x="total_bill", y="tip")
sns.despine()
plt.show()
```
