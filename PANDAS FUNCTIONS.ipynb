```python
#df.dropna():It removes rows where values are missing.
df = pd.DataFrame({
    'Name': ['A', 'B', 'C'],
    'Marks': [80, None, 70]
})

print(df.dropna())
```

```python
#df.drop():removes rows or columns based on labels.
df = pd.DataFrame({
    'Name': ['A', 'B', 'C'],
    'Marks': [80, 90, 70]
})
print(df.drop('Marks', axis=1)) 
```

```python
#df.replace():Replaces values in the DataFrame with specified values.
df = pd.DataFrame({
    'Name': ['A', 'B', 'C'],
    'Marks': [80, 90, 70]
})
print(df.replace(80, 85))
```

```python
#df.groupby():groups the DataFrame by one or more columns and allows for aggregation operations.
df = pd.DataFrame({
    'Name': ['R', 'P', 'A', 'S'],
    'Marks': [80, 90, 70, 85]
})
grouped = df.groupby('Name')
print(grouped.mean())
```

```python
#df.rename():Renames columns or indexes in the DataFrame.
df = pd.DataFrame({
    'Name': ['ANU', 'SAI', 'CHINNA'],
    'Marks': [70, 88, 99]
})
df_renamed = df.rename(columns={'Name': 'Student Name', 'Marks': 'Score'})
print(df_renamed)
```

```python
#df.mean():Calculates the mean of numeric columns in the DataFrame.
df = pd.DataFrame({
    'Name': ['ANU', 'SAI', 'CHINNA'],
    'Marks': [70, 88, 99]
})
mean_marks = df['Marks'].mean()
print("Mean of Marks:", mean_marks)
```

```python
#df.median():Calculates the median of dataframe columns.
print(df['Marks'].median())
```

```python
#df.mode():It finds the most repeated value.
print(df['Marks'].mode())
```

```python
#df.sum():It calculates the total of numerical values.
df = pd.DataFrame({
    'total points': [80, 90, 70]
})

print(df['total points'].sum())
```

```python
#df.max(): It returns the largest value.
print(df['total points'].max())
```

```python
#df.min(): It returns the smallest value.
print(df['total points'].min())
```

```python
#df.count(): It counts non-missing values in each column.
print(df['total points'].count())
```

```python
#df.var():It calculates the variance of numerical data.
print(df['total points'].var())
```

```python
#df.std(): It calculates the standard deviation of numerical data.
print(df['total points'].std())
```

```python
#df.apply():It applies a function to each value, row, or column.
import pandas as pd

df = pd.DataFrame({
    'total points': [56, 67, 87]
})

df['total points'] = df['total points'].apply(lambda x: x + 5)

print(df)
```

```python
#df.corr():It calculates the correlation between numerical columns.
df = pd.DataFrame({
    'student': [1, 2, 3, 4],
    'total points': [56, 67, 87, 98]
})

print(df.corr())
```

```python
#df.map():It applies a function to every value in a Series.
df = pd.DataFrame({
    'total points': [56, 67, 87]
})

print(df['total points'].map(lambda x: x * 2))
```

```python
#df.sort_values():It sorts rows based on one or more columns.
df = pd.DataFrame({
    'Name': ['Anu', 'Sai', 'chinna'],
    'total points': [80, 95, 70]
})

print(df.sort_values('total points', ascending=False))
```

```python
#pd.concat():It combines two or more DataFrames vertically or horizontally.
df1 = pd.DataFrame({
    'Name': ['sai', 'chinna']
})

df2 = pd.DataFrame({
    'Name': ['jaya', 'Anu']
})

print(pd.concat([df1, df2], ignore_index=True))
```

```python
#df.sort_index(): It arranges rows according to their index values.
df = pd.DataFrame({
    'Marks': [80, 90, 70]
}, index=[2, 0, 1])

print(df.sort_index())
```

```python
#df.join():It joins two DataFrames using their index.
df1 = pd.DataFrame({
    'Name': ['anu', 'Sai']
}, index=[1, 2])

df2 = pd.DataFrame({
    'Marks': [80, 90]
}, index=[1, 2])

print(df1.join(df2))
```

```python
#pd.merge():It combines dataframes using a common column.
df1 = pd.DataFrame({
    'ID': [1, 2],
    'Name': ['anu', 'Sai']
})

df2 = pd.DataFrame({
    'ID': [1, 2],
    'Marks': [809, 920]
})

print(pd.merge(df1, df2, on='ID'))
```

```python
#df.all(): It checks whether every value is True.
print(df.all())
```

```python
#df.copy():It creates an separate dataframe so changes don't directly affect the original.
df = pd.DataFrame({
    'Marks': [80, 90]
})

new_df = df.copy()

print(new_df)
```

```python
#df.shift():It moves data by a specified number of rows.
df = pd.DataFrame({
    'Marks': [622, 860, 790]
})

print(df['Marks'].shift(1))
```

```python
#df.any():It checks whether at least one value is True in a row or column.
df = pd.DataFrame({
    'A': [True, False, False],
    'B': [False, False, True]
})

print(df.any())
```

```python
#df.all(): It checks whether every value is True.
print(df.all())
```
