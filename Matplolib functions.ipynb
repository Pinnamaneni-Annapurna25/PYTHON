```python
import matplotlib.pyplot as plt
```

```python
#plt.plot():it draws a line plot of the given data points. It takes in x and y coordinates as arguments and can also accept additional parameters to customize the appearance of the plot, such as color, line style, and markers.
x=[30,40,50,60,70]
y=[10,20,30,40,50]
plt.plot(x, y)
```

```python
#plt.scatter():it creates a scatter plot of the given data points. It takes in x and y coordinates as arguments and can also accept additional parameters to customize the appearance of the plot, such as color, marker style, and size.
x=[30,40,50,60,70]
y=[10,20,30,40,50]
plt.scatter(x, y)
```

```python
#plt.bar():it creates a bar plot of the given data points. It takes in x and y coordinates as arguments and can also accept additional parameters to customize the appearance of the plot, such as color, width, and orientation.
x=[30,40,50,60,70]
y=[10,20,30,40,50]
plt.bar(x, y)
```

```python
#plt.subplots():it creates a figure and a set of subplots. It can be used to create multiple plots in a single figure and can also accept additional parameters to customize the appearance of the subplots, such as the number of rows and columns, the size of the figure, and the spacing between subplots.
x=[30,40,50,60,70]
y=[10,20,30,40,50]
fig, ax = plt.subplots()
```

```python
#plt.xlabel():it sets the label for the x-axis of the plot. It takes in a string as an argument, which will be displayed as the label for the x-axis. Additional parameters can also be used to customize the appearance of the label, such as font size and color.
x=[30,40,50,60,70]
y=[10,20,30,40,50]
plt.plot(x, y)
plt.xlabel('X-axis label')
```

```python
#plt.ylabel():it sets the label for the y-axis of the plot. It takes in a string as an argument, which will be displayed as the label for the y-axis. Additional parameters can also be used to customize the appearance of the label, such as font size and color.
x=[30,40,50,60,70]
y=[10,20,30,40,50]
plt.plot(x, y)
plt.ylabel('Y-axis label')
```

```python
#plt.grid(True):it adds a grid to the plot. It takes in a boolean value as an argument, which determines whether the grid should be displayed or not. Additional parameters can also be used to customize the appearance of the grid, such as color, line style, and width.

plt.plot([15, 24, 83], [10, 20, 30])
plt.grid()
plt.show()


```

```python
#plt.hist(): It shows the distribution/frequency of numerical data.
values = [70, 67, 90, 30]
labels = ["AB", "BC", "CD", "DC"]

plt.pie(values, labels=labels)
plt.show()
```

```python
#plt.legend():It displays labels for multiple plots.
x = [1, 2, 3]

plt.plot(x, [10, 20, 30], label="MARKS")
plt.plot(x, [5.3, 6.5, 8.5], label="POINTS")

plt.legend()
plt.show()
```

```python
#plt.figure():It creates a new figure/window for plotting.
plt.figure(figsize=(10, 9))
plt.plot([31, 28, 63], [40, 65, 50])
plt.show()
```

```python
import io
from IPython.display import Image, display

#plt.savefig(): It saves the graph as an image file.
plt.plot([1, 2, 3], [10, 20, 30])
buf = io.BytesIO()
plt.savefig(buf, format='png', bbox_inches='tight')
buf.seek(0)
display(Image(data=buf.getvalue()))
plt.show()
```

```python
#plt.figure():It creates a new figure/window for plotting.
plt.figure(figsize=(8, 5))
plt.plot([1, 2, 3], [10, 20, 30])
plt.show()
```

```python
#plt.close(): It closes the current figure.
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [10, 20, 30])
plt.close()
```

```python
#plt.contour():It creates contour lines for 3-dimensional data.
import numpy as np

x = np.linspace(-23, 32, 50)
y = np.linspace(-13, 43, 58)

X, Y = np.meshgrid(x, y)
Z = X**2 + Y**2

plt.contour(X, Y, Z)
plt.show()
```
